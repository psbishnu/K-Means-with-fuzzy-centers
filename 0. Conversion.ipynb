```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

# Load dataset
df = pd.read_csv("car_evaluation.csv")

# Convert categorical columns to numerical values
label_encoders = {}

for col in df.columns:
    if df[col].dtype == "object" or df[col].dtype.name == "category":
        le = LabelEncoder()
        df[col] = le.fit_transform(df[col].astype(str))
        label_encoders[col] = le

# Save converted dataset
df.to_csv("converted_numerical_dataset.csv", index=False)

print("Categorical columns converted successfully.")
print("Converted dataset saved as converted_numerical_dataset.csv")
print("Encoded columns:", list(label_encoders.keys()))
```
