```python
import pandas as pd

# ==============================
# Configuration
# ==============================


DATASET_PATH = "car.csv"  # Example: "heart (2), iris(3) car (4), glass (6)"
LABEL_COLUMN = "acceptability"          # Example: "target"; keep None if no class label heart: target, iris: variety, car:acceptability, glass:Type

# ==============================
# Load dataset
# ==============================
df = pd.read_csv(DATASET_PATH)

# Number of samples and features
n_samples = len(df)

if LABEL_COLUMN is not None and LABEL_COLUMN in df.columns:
    n_features = len(df.columns) - 1
    n_classes = df[LABEL_COLUMN].nunique()

    print("=" * 40)
    print(f"Number of Samples : {n_samples}")
    print(f"Number of Features: {n_features}")
    print(f"Number of Classes : {n_classes}")
    print("\nClass Distribution:")
    print(df[LABEL_COLUMN].value_counts().sort_index())
    print("=" * 40)

else:
    n_features = len(df.columns)

    print("=" * 40)
    print(f"Number of Samples : {n_samples}")
    print(f"Number of Features: {n_features}")
    print("Class labels not specified.")
    print("=" * 40)
```
